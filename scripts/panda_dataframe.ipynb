```python
import pandas as pd
from pathlib import Path
from urllib.request import urlretrieve

url = "https://raw.githubusercontent.com/pandas-dev/pandas/main/pandas/tests/io/data/csv/tips.csv"
filename = Path.cwd() / "tips.csv"

urlretrieve(url, filename)
print(f"Downloaded to: {filename}")

df = pd.read_csv(filename)
print(df)

average_total_bill = df["total_bill"].mean()
df["total_bill"] = df["total_bill"].fillna(average_total_bill)


df_new = df.drop(columns=["sex"])
df_new.to_csv("output.csv", index=False)
print(df_new)
print(average_total_bill)
```
