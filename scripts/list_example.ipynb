```python
number = [1,2,3,4,5]
print(number)
```

```python
new_list = [item * item for item in number if item % 2 == 0]
new_list
```

```python
numbers = [1, 35, 25, 50, 100]
new_list = [number / 50 for number in numbers]
new_list
```

```python
numbers = [1, 35, 25, 50, 100]
mult = lambda x: x * x
print(mult)
```

```python
numbers = [1, 35, 25, 50, 100]

mult = lambda x: x * x
new_list = list(map(mult, numbers))

print(new_list)
```

```python
numbers = [1, 35, 25, 50, 100]

mult = lambda x: x / 25 == 0
new_list = list(map(mult, numbers))

print(new_list)
```

```python
numbers = [1, 35, 25, 50, 100]

new_list = list(filter(lambda number: number % 25 == 0, numbers))

print(new_list)
```

```python
numbers = [1, 35, 25, 50, 100]

from functools import reduce
new_list = reduce(sum(numbers), numbers)

print(new_list)
```

```python
from functools import reduce

numbers = [1, 35, 25, 50, 100]

final_sum = reduce(lambda total, number: total + number, numbers)

print(final_sum)
```

```python
actual = [1,0,1,1,0]
predicted = [1,1,0,1,0]

mapped = zip(actual, predicted)

print(list(mapped))
```

```python
actuals = [1, 0, 1, 1, 0]
predicted = [1, 1, 0, 1, 0]

for actual, prediction in zip(actuals, predicted):
    print(f"actual: {actual}, predicted: {prediction}")
```

```python
actuals = [1, 0, 1, 1, 0]
predicted = [1, 1, 0, 1, 0]

match = zip(actuals, predicted)

for i, (a, p) in enumerate(match):
    if a != p:
        print(f"predicted: {p} are not correct")
```

```python
actuals = [1, 0, 1, 1, 0]
predicted = [1, 1, 0, 1, 0]

incorrect = sum(a != p for a, p in zip(actuals, predicted))
print(f"{incorrect} predictions are incorrect")
```
